Requires `main.py` to have been run first.

```python
import os
import sys

import matplotlib.pyplot as plt

# assumes Jupyter's cwd is this notebook's own directory (analysis/), so the
# repo root - and config.py - is one level up
REPO_ROOT = os.path.abspath('..')
sys.path.insert(0, REPO_ROOT)
import config

# config.py's OUTPUT_DIR/CAJAL_OUTPUT_DIR are relative to the repo root, not
# this notebook's own cwd - re-anchor them to an absolute path once, here
config.OUTPUT_DIR       = os.path.join(REPO_ROOT, config.OUTPUT_DIR)
config.CAJAL_OUTPUT_DIR = os.path.join(REPO_ROOT, config.CAJAL_OUTPUT_DIR)

from helpers.load_data import load_full_dataset
from helpers.density_plots import export_kde_by_frame
```

## Plot: UMAP with per-subclass kernel density, vem only

One combined, multi-page PDF (one page per frame) and one SVG per frame, written to
`output/plots/03_vem_subclasses_over_time/pdfs/` and `.../svgs/`.

```python
SUBCLASS_COLORS = {
    'dying':       'tab:red',
    'persisting':  'tab:purple',
    'slow_prolif': 'tab:orange',
    'fast_prolif': 'tab:green',
}
NOTEBOOK_PLOTS_DIR = os.path.join(config.OUTPUT_DIR, 'plots', '03_vem_subclasses_over_time')

def process_modality(modality):
    '''
    Load this modality's embedding + metadata, restrict to the vem
    condition, and export one PDF/SVG pair per frame with a KDE
    contour per subclass.

    Inputs:
        modality: 'cell' or 'nuc'
    Outputs:
        vem_df: the vem-only subset for this modality (in-memory only)
    '''
    df = load_full_dataset(modality)
    vem_df = df[df['condition'] == config.VEM_CONDITION]
    print(f"[{modality}] {len(vem_df)} vem cells across "
          f"{vem_df['subclass'].nunique()} subclasses, {vem_df['frame'].nunique()} frames")

    export_kde_by_frame(vem_df, 'subclass', SUBCLASS_COLORS, NOTEBOOK_PLOTS_DIR,
                         modality, title_prefix=f'{modality} - vem subclasses')

    return vem_df
```

## Run: cell morphology

```python
cell_vem_df = process_modality('cell')
cell_vem_df.head()
```

## Run: nuclear morphology
