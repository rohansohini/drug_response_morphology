# Proliferators: vem vs. naive, Over Time (Cell + Nucleus)

Same shared UMAP embedding as `01_umap_by_condition_over_time.ipynb`,
restricted to proliferators only (`m_label == 3` - slow and fast
proliferator sub-classes pooled together), comparing vem vs. naive at
each overlapping frame.

Requires `main.py` to have been run first.

```python
import os
import sys

import matplotlib.pyplot as plt

# assumes Jupyter's cwd is this notebook's own directory (analysis/), so the
# repo root - and config.py - is one level up
REPO_ROOT = os.path.abspath('..')
sys.path.insert(0, REPO_ROOT)
import config

# config.py's OUTPUT_DIR/CAJAL_OUTPUT_DIR are relative to the repo root, not
# this notebook's own cwd - re-anchor them to an absolute path once, here
config.OUTPUT_DIR       = os.path.join(REPO_ROOT, config.OUTPUT_DIR)
config.CAJAL_OUTPUT_DIR = os.path.join(REPO_ROOT, config.CAJAL_OUTPUT_DIR)

from helpers.load_data import load_full_dataset
from helpers.density_plots import export_kde_by_frame
```

## Plot: UMAP with per-condition kernel density, proliferators only

One PDF and one matching SVG per frame, written to
`output/plots/02_proliferators_vem_vs_naive/pdfs/` and `.../svgs/`.

```python
CONDITION_COLORS = {'naive': 'tab:blue', 'A1_1uM_vem': 'tab:red'}
PROLIF_M_LABEL = config.PROLIF_M_LABEL
NOTEBOOK_PLOTS_DIR = os.path.join(config.OUTPUT_DIR, 'plots', '02_proliferators_vem_vs_naive')

def process_modality(modality):
    '''
    Load this modality's embedding + metadata, restrict to
    proliferators (m_label == config.PROLIF_M_LABEL) in both
    conditions, and export one PDF/SVG pair per overlapping frame.

    Inputs:
        modality: 'cell' or 'nuc'
    Outputs:
        prolif_df: the proliferator-only subset for this modality
                   (in-memory only)
    '''
    df = load_full_dataset(modality)
    prolif_df = df[df['m_label'] == PROLIF_M_LABEL]
    print(f"[{modality}] {len(prolif_df)}/{len(df)} cells are proliferators "
          f"(m_label == {PROLIF_M_LABEL})")

    export_kde_by_frame(prolif_df, 'condition', CONDITION_COLORS, NOTEBOOK_PLOTS_DIR,
                         modality, title_prefix=f'{modality} - proliferators only')

    return prolif_df
```

## Run: cell morphology

```python
cell_prolif_df = process_modality('cell')
cell_prolif_df.head()
```

## Run: nuclear morphology
